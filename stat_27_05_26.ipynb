```python
n=300        #sample size
Xbar=1800    #mean
stdev = 70
cl = 0.95
```

```python
norm.ppf((1-cl)/2)
```

```python
z = np.abs(norm.ppf((1-cl)/2))
```

```python
z
```

```python
t = np.abs(t.ppf((1-cl)/2,n-1))             #t distribution needs degrres of freedom so n-1
```

```python
t
```

```python
low_est = Xbar - t*stdev/n**0.5
```

```python
upper_est = Xbar + t*stdev/n**0.5
```

```python
size = upper_est - low_est
```
