# ***insert a col grade , if 10<=quanty<=50 -A , 50<quanty<100 -B , 100<=Qnty -c***

```python
# step 1 : we have to create a list , a loop if elif else: fill up that list , once completed add it to the column.
```

```python
grd = []
```

#***Adding a column by generating values based on an existing column***

```python
for i in df.Quantity:
  if i<=50:
    grd.append("A")
  elif i<100:
    grd.append("B")
  else:
    grd.append("C")
df['Grade'] = grd
```
